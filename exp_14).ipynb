```python
# Experiment 14 — Feature Scaling

# Import the required libraries
import numpy as np
import pandas as pd

# Import preprocessing tools from scikit-learn
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from google.colab import files

uploaded = files.upload()

# Read the dataset
df = pd.read_csv("pre-process_datasample.csv")

# Display the original dataset
print("Original Dataset:")
print(df)

# Fill the missing Country value with the most frequent country
print("\n\nDataset after filling the missing country value with mode:")
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])
print(df)

print("\n\nDataset after imputing the dataset with missing age and salary with np.nan to fill it with mean:")
# Create an imputer to replace missing Age values with the mean
age = SimpleImputer(strategy="mean", missing_values=np.nan)

# Create an imputer to replace missing Salary values with the mean
salary = SimpleImputer(strategy="mean", missing_values=np.nan)


# Separate input features and target label iloc selects all rows and columns df.iloc[:, :-1]	All rows, all columns except the last
# df.iloc[:, -1]	All rows, only the last column
# .values	Convert the selected data into a NumPy array
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

# Calculate the mean Age and Salary fit calculates and remembers the mean ignoring missing values of the selected column
# features found above gives the required rows and columns
age.fit(features[:, [1]])
salary.fit(features[:, [2]])

# Replace missing values in Age and Salary
# transform() — uses fit's mean to replace missing Age values.
features[:, [1]] = age.transform(features[:, [1]])
features[:, [2]] = salary.transform(features[:, [2]])


# Display the data after handling missing values
print("\n\nData after Missing-Value Handling with mean:")
print(features)

# Convert the categorical Country column into numerical columns
oh = OneHotEncoder(sparse_output=False)
Country = oh.fit_transform(features[:, [0]])

# Display the one-hot encoded Country values
print("\n\nOne-Hot Encoded Country:")
print(Country)

# Combine encoded Country with Age and Salary axis joins row by row or col by col
# axis=0 → refers to rows.
# axis=1 → refers to columns.
final_set = np.concatenate(
    (Country, features[:, [1, 2]]),
    axis=1
)

print("\n\nFinal Feature Set:")
print(final_set)


# StandardScaler — changes values so they have a mean of approximately 0 and a standard deviation of 1.
# MinMaxScaler — scales values to a range between 0 and 1.


# Apply StandardScaler to standardize the features
sc = StandardScaler()
sc.fit(final_set)

# Transform the data to have approximately zero mean
# and unit variance in each feature
feat_standard_scaler = sc.transform(final_set)

print("\n\nStandard Scaled Features:")
print(feat_standard_scaler)

print("\n\nMean of each column:")
print(feat_standard_scaler.mean(axis=0))

print("\n\nStandard deviation of each column:")
print(feat_standard_scaler.std(axis=0))

# Apply MinMaxScaler to scale values between 0 and 1
mms = MinMaxScaler(feature_range=(0, 1))
mms.fit(final_set)

# Transform the data
feat_minmax_scaler = mms.transform(final_set)

print("\n\nMin-Max Scaled Features:")
print(feat_minmax_scaler)
```
