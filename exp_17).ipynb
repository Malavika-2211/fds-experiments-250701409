```python
# Exp No:17 Experiment to understand Linear Regression for a given
# data set.
# Description: Understand the Linear regression for the dataset given.
# Step 1: Import the required libraries
import numpy as np
import pandas as pd
from google.colab import files
uploaded=files.upload()

# Step 2: Read the dataset
df = pd.read_csv('Salary_data.csv')
print()
print(df)

# Step 3:
print("\n\nDisplay dataset information:\n\n")
df.info()

# Step 4: Remove rows containing missing values
df.dropna(inplace=True)

# Step 5:
print("\n\nDisplay dataset information after removing rows containing missing values:\n\n")
df.info()

# Step 6:
print("\n\nDisplay statistical information:\n\n")
df.describe()

# Step 7: Separate features and labels
features = df.iloc[:, [0]].values
label = df.iloc[:, [1]].values

# Step 8: Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=0
)

# Step 9: Import and create the Linear Regression model
from sklearn.linear_model import LinearRegression

model = LinearRegression()

# Step 10: Train the model
model.fit(x_train, y_train)

# Step 11: Check the training score
model.score(x_train, y_train)

# Step 12: Check the testing score
model.score(x_test, y_test)

# Step 13: Display the model coefficient
print(model.coef_)

# Step 14: Display the model intercept
print(model.intercept_)

# Step 15: Save the trained model
import pickle

pickle.dump(model, open('SalaryPred.model', 'wb'))

# Step 16: Load the saved model
model = pickle.load(open('SalaryPred.model', 'rb'))

# Step 17: Enter years of experience and predict salary
yr_of_exp = float(input("Enter Years of Experience: "))

yr_of_exp_NP = np.array([[yr_of_exp]])

Salary = model.predict(yr_of_exp_NP)

# Step 18: Print the estimated salary
print("\n\nEstimated Salary for {} years of experience is {}"
      .format(yr_of_exp, Salary))
```
