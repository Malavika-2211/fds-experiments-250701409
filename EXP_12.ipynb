```python
# EXP NO. 12
# OUTLIER DETECTION USING NUMPY AND SEABORN

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# CREATE ARRAY
array = np.array([
    27, 50, 44, 6, 58, 61, 23, 86, 67, 20,
    75, 7, 79, 61, 90, 54
])

print("ORIGINAL ARRAY:")
print(array)


# FIND 25TH PERCENTILE
Q1 = np.percentile(array, 25)
print("\n25TH PERCENTILE:")
print(Q1)


# FIND 75TH PERCENTILE
Q3 = np.percentile(array, 75)
print("\n75TH PERCENTILE:")
print(Q3)


# OUTLIER DETECTION FUNCTION
def outlierDetection(array):

    sorted_array = sorted(array)

    Q1, Q3 = np.percentile(sorted_array, [25, 75])

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    return lower_limit, upper_limit


# FIND LOWER AND UPPER LIMIT
lr, ur = outlierDetection(array)

print("\nLOWER LIMIT:")
print(lr)

print("\nUPPER LIMIT:")
print(ur)


# HISTOGRAM
sns.histplot(array)

plt.xlabel("VALUES")
plt.ylabel("COUNT")
plt.title("HISTOGRAM OF ORIGINAL DATA")

plt.show()


# DISTRIBUTION PLOT
sns.kdeplot(array, fill=True)

plt.xlabel("VALUES")
plt.ylabel("DENSITY")
plt.title("DISTRIBUTION OF ORIGINAL DATA")

plt.show()


# REMOVE OUTLIERS
new_array = array[
    (array > lr) & (array < ur)
]

print("\nARRAY AFTER REMOVING OUTLIERS:")
print(new_array)


# FINAL ARRAY
final_array = new_array

print("\nFINAL ARRAY:")
print(final_array)


# HISTOGRAM AFTER REMOVING OUTLIERS
sns.histplot(final_array)

plt.xlabel("VALUES")
plt.ylabel("COUNT")
plt.title("HISTOGRAM AFTER REMOVING OUTLIERS")

plt.show()


# FINAL DISTRIBUTION PLOT
sns.kdeplot(final_array, fill=True)

plt.xlabel("VALUES")
plt.ylabel("DENSITY")
plt.title("FINAL DISTRIBUTION AFTER REMOVING OUTLIERS")

plt.show()
```
