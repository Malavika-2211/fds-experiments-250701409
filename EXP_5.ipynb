```python
# 5 DATA COLLECTION AND INITIAL EXPLORATION

# Step 1: Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Upload the CSV file
from google.colab import files

uploaded = files.upload()

# Step 3: Load the dataset
diabetes_df = pd.read_csv("diabetes.csv")

# Step 4: Display the first 5 rows
print("\nFirst 5 rows:")
print(diabetes_df.head())

# Step 5: Display dataset information
print("\nDataset Information:")
print(diabetes_df.info())

# Step 6: Display summary statistics
print("\nSummary Statistics:")
print(diabetes_df.describe())

# Step 7: Display the number of rows and columns
print("\nDataset Shape:")
print(diabetes_df.shape)

# Step 8: Check for missing values
print("\nMissing Values:")
print(diabetes_df.isnull().sum())

# Step 9: Visualize distribution of each feature
print("\nHistograms:")
diabetes_df.hist(bins=50, figsize=(20, 15))
plt.tight_layout()
plt.show()

# Step 10: Pairplot to visualize relationships
print("\nPairplot:")
sns.pairplot(diabetes_df)
plt.show()
```
