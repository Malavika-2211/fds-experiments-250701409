```python
# EXPERIMENT 13: HANDLING MISSING VALUES AND INAPPROPRIATE DATA IN A GIVEN DATA SET

# Step 1: Import the required libraries
import numpy as np
import pandas as pd
from google.colab import files

# Step 2: Upload the dataset
uploaded = files.upload()

# Step 3: Read the CSV file
filename = next(iter(uploaded))
df = pd.read_csv(filename)

# Display the original dataset
print("\n\nOriginal Dataset:")
display(df)

# Step 4: Check for duplicate rows
print("\n\nDuplicate Rows:")
print(df.duplicated())

# Step 5: Display dataset information
print("\nDataset Information:")
df.info()

# Step 6: Remove duplicate rows
df.drop_duplicates(inplace=True)

print("\n\nNumber of rows after removing duplicates:")
print(len(df))

# Step 7: Reset the index
index = np.array(list(range(0, len(df))))
df.set_index(index, inplace=True)

# Step 8: Remove the duplicate Age_Group.1 column if it exists
if "Age_Group.1" in df.columns:
    df.drop(["Age_Group.1"], axis=1, inplace=True)

# Step 9: Replace inappropriate negative values with NaN
if "CustomerID" in df.columns:
    df.loc[df["CustomerID"] < 0, "CustomerID"] = np.nan

if "Bill" in df.columns:
    df.loc[df["Bill"] < 0, "Bill"] = np.nan

if "EstimatedSalary" in df.columns:
    df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan

# Step 10: Handle inappropriate NoOfPax values
# Values below 1 or above 20 are treated as missing
if "NoOfPax" in df.columns:
    df.loc[
        (df["NoOfPax"] < 1) | (df["NoOfPax"] > 20),
        "NoOfPax"
    ] = np.nan

# Step 11: Display unique age groups
if "Age_Group" in df.columns:
    print("\nUnique Age Groups:")
    print(df["Age_Group"].unique())

# Step 12: Display the cleaned dataset
print("\n\nCleaned Dataset:")
display(df)

# Step 13: Check missing values in each column
print("\n\nMissing Values in Each Column:")
print(df.isnull().sum())

# Step 14: Display final dataset information
print("\n\nFinal Dataset Information:")
df.info()
```
