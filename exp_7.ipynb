```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate a population
population_mean = 50
population_std = 10
population_size = 100000

population = np.random.normal(
    population_mean,
    population_std,
    population_size
)

# Step 2: Define sample sizes
sample_sizes = [30, 50, 100]

# Number of random samples for each sample size
num_samples = 1000

# Store sample means
sample_means = {}

for size in sample_sizes:
    sample_means[size] = []

    for _ in range(num_samples):
        # Randomly select a sample from the population
        sample = np.random.choice(
            population,
            size=size,
            replace=False
        )

        # Calculate the sample mean
        sample_means[size].append(np.mean(sample))


# Step 3: Plot sampling distributions
plt.figure(figsize=(12, 8))

for i, size in enumerate(sample_sizes):

    plt.subplot(len(sample_sizes), 1, i + 1)

    plt.hist(
        sample_means[size],
        bins=30,
        alpha=0.7,
        label=f'Sample Size {size}'
    )

    # Population mean
    plt.axvline(
        population_mean,
        color='red',
        linestyle='dashed',
        linewidth=1.5,
        label='Population Mean'
    )

    plt.title(f'Sampling Distribution (Sample Size {size})')
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')
    plt.legend()

plt.tight_layout()
plt.show()
```
