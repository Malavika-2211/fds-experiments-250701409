```python
#EXPERIMENT TO UNDERSTAND MATPLOTLIB LIBRARY USING PYTHON LIBRARIES THROUGH VISUALIZATION
# Exp No. 9
# Visualization using Matplotlib with Iris Dataset

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Step 1: Load Iris dataset
iris = load_iris()

# Convert dataset into DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add target/species column
df['Species'] = iris.target
df['Species Name'] = df['Species'].map({
    0: 'Setosa',
    1: 'Versicolor',
    2: 'Virginica'
})

# Display first five records
print("First Five Rows:")
print(df.head())

# Display dataset information
print("\nDataset Description:")
print(df.describe())


# --------------------------------------------------
# 1. Scatter Plot
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
    c=df['Species'],
    alpha=0.7
)

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Iris Dataset - Sepal Length vs Sepal Width')
plt.grid(True)

plt.show()


# --------------------------------------------------
# 2. Line Plot
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    df.index,
    df['petal length (cm)']
)

plt.xlabel('Sample Index')
plt.ylabel('Petal Length (cm)')
plt.title('Petal Length Across Iris Samples')
plt.grid(True)

plt.show()


# --------------------------------------------------
# 3. Histogram
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df['sepal length (cm)'],
    bins=15,
    alpha=0.7
)

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Sepal Length')

plt.show()


# --------------------------------------------------
# 4. Bar Chart
# --------------------------------------------------

species_count = df['Species Name'].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    species_count.index,
    species_count.values
)

plt.xlabel('Species')
plt.ylabel('Number of Samples')
plt.title('Number of Samples in Each Iris Species')

plt.show()


# --------------------------------------------------
# 5. Box Plot
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        df['sepal length (cm)'],
        df['sepal width (cm)'],
        df['petal length (cm)'],
        df['petal width (cm)']
    ],
    labels=[
        'Sepal Length',
        'Sepal Width',
        'Petal Length',
        'Petal Width'
    ]
)

plt.ylabel('Measurement (cm)')
plt.title('Iris Dataset - Feature Distribution')

plt.show()


# --------------------------------------------------
# 6. Multiple Feature Visualization
# --------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    df.index,
    df['sepal length (cm)'],
    label='Sepal Length'
)

plt.plot(
    df.index,
    df['petal length (cm)'],
    label='Petal Length'
)

plt.xlabel('Sample Index')
plt.ylabel('Length (cm)')
plt.title('Sepal Length and Petal Length')
plt.legend()
plt.grid(True)

plt.show()
```
