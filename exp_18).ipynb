```python
# Experiment 18 — KNN Algorithm
# Experiment: K-Nearest Neighbors (KNN) Classification

# Step 1: Import the required libraries
import numpy as np
import pandas as pd

# Import the function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import the KNN classification algorithm
from sklearn.neighbors import KNeighborsClassifier

# Import functions to evaluate the model
from sklearn.metrics import confusion_matrix, classification_report

# Import Google Colab file upload function
from google.colab import files

# Upload the Iris dataset from your computer
uploaded = files.upload()

# Step 2: Read the Iris dataset
df = pd.read_csv("Iris.csv")

# Step 3: Display information about the dataset
print("\n\nDataset information:")
df.info()

# Display the number of samples belonging to each flower variety
print("\n\nNumber of samples belonging to each flower variety:")
print(df["variety"].value_counts())

# Display the first five rows of the dataset
print("\n\nFirst five rows of the dataset:")
print(df.head())

# Step 4: Separate input features and target labels
# Select all rows and all columns except the last column
# Example: Features may include sepal length, sepal width,
# petal length, and petal width
features = df.iloc[:, :-1].values

# Select the fifth column (index 4) containing flower varieties
# Example: Setosa, Versicolor, or Virginica
label = df.iloc[:, 4].values

# Step 5: Split the dataset into training and testing sets
# 80% of the data is used for training
# 20% of the data is used for testing
# random_state=42 makes the split reproducible
xtrain, xtest, ytrain, ytest = train_test_split(
    features,
    label,
    test_size=0.2,
    random_state=42
)

# Step 6: Create the KNN classifier with 5 nearest neighbors
# Example: The model looks at the 5 closest training flowers
# and predicts the variety based on their labels
model_KNN = KNeighborsClassifier(n_neighbors=5)

# Step 7: Train the KNN model
# The model learns from the training measurements and labels
model_KNN.fit(xtrain, ytrain)

# Step 8: Display the training accuracy
# Example: 0.95 means 95% of training samples were classified correctly
print("\n\nAccuracy of the model on training data:")
print(model_KNN.score(xtrain, ytrain))

# Display the testing accuracy
# This evaluates the model on data kept aside during training
print("\n\nAccuracy of the model on testing data:")
print(model_KNN.score(xtest, ytest))

# Step 9: Predict the varieties of testing flowers
# Example: Predictions may include Setosa or Virginica
y_pred = model_KNN.predict(xtest)

# Display the confusion matrix
# Compares actual testing labels with predicted labels
# Diagonal values represent correct classifications
print("\n\nConfusion Matrix:")
print(confusion_matrix(ytest, y_pred))

# Step 10: Display the classification report
# Precision: How many predicted samples of a variety were correct
# Recall: How many actual samples of a variety were identified
# F1-score: Combined measure of precision and recall
# Support: Number of actual testing samples in each variety
print("\n\nClassification Report:")
print(classification_report(ytest, y_pred))
```
