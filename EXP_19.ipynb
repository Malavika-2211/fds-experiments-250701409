```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from google.colab import files
uploaded = files.upload()
# Read the dataset
df = pd.read_csv("Social_Network_Ads.csv")
print("DataFrame:")
print(df)
# Display the first 5 rows
print("\n\nDataframe's first 5 rows:")
print(df.head())

# Separate features and label
features = df.iloc[:, [2, 3]].values
label = df.iloc[:, 4].values

# Split the dataset into training and testing data
x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=37
)

# Create the Logistic Regression model
finalModel = LogisticRegression()

# Train the model
finalModel.fit(x_train, y_train)

# Print training accuracy
print("\n\nTraining Accuracy:")
print(finalModel.score(x_train, y_train))

# Print testing accuracy
print("\n\nTesting Accuracy:")
print(finalModel.score(x_test, y_test))
#Note: The accuracy and classification report depend on the contents of your dataset.
# to evaluate only the test data:
print("\n\nclassfication report for only test data")
print(classification_report(y_test, finalModel.predict(x_test)))

# Print classification report which evaluates the entire dataset
print("\n\nClassification Report:")
print(classification_report(label, finalModel.predict(features)))
```
